```python
import numpy as np 
import matplotlib.pyplot as plt 
import math,copy
```

```python
def compute_cost_lr_reg(X,y,w,b,lambda_=1):
    m,n = X.shape
    cost = 0.
    for i in range(m):
        f_wb_i = np.dot(X[i],w)+b
        cost+= ((f_wb_i-y[i])**2)
    cost/=2*m
    reg_cost = 0.
    for j in range(n):
        reg_cost+= (w[j]**2)
    reg_cost = (lambda_/(2*m))*reg_cost
    total_cost = cost + reg_cost
    return total_cost
```

```python
np.random.seed(1)
X_train = np.random.rand(5,6)
y_train = np.array([0,1,0,1,0])
w_tmp = np.random.rand(X_train.shape[1]).reshape(-1,)-0.5
b_tmp = 0.5
lambda_tmp = 0.7
cost_tmp = compute_cost_lr_reg(X_train,y_train,w_tmp,b_tmp,lambda_tmp)
print("Regularized Cost:",cost_tmp)
```

```python
def sigmoid(z):
    g = 1/(1+np.exp(-z))
    return g
```

```python
def compute_cost_logreg_reg(X,y,w,b,lambda_=1):
    m,n = X.shape
    cost = 0.
    for i in range(m):
        z_i = np.dot(X[i],w)+b
        f_wb_i = sigmoid(z_i)
        cost+= -y[i]*np.log(f_wb_i)-(1-y[i])*np.log(1-f_wb_i)
    cost/=m
    reg_cost = 0.
    for j in range(n):
        reg_cost+= (w[j]**2)
    reg_cost = (lambda_/(2*m))*reg_cost
    total_cost = cost + reg_cost
    return total_cost
```

```python
np.random.seed(1)
X_train = np.random.rand(5,6)
y_train = np.array([0,1,0,1,0])
w_tmp = np.random.rand(X_train.shape[1]).reshape(-1,)-0.5
b_tmp = 0.5
lambda_tmp = 0.7
cost_tmp = compute_cost_logreg_reg(X_train,y_train,w_tmp,b_tmp,lambda_tmp)
print("Regularized Cost:",cost_tmp)
```

```python
def compute_gradient_linear_reg(X,y,w,b,lambda_=1):
    m,n=X.shape
    dj_dw = np.zeros(n)
    dj_db = 0.
    for i in range(m):
        err = (np.dot(X[i],w)+b)-y[i]
        for j in range(n):
            dj_dw[j]+= err*X[i,j]
        dj_db+= err
    dj_dw/=m
    dj_db/=m
    for j in range(n):
        dj_dw[j]+= (lambda_/m)*w[j]
    return dj_dw,dj_db
```

```python
np.random.seed(1)
X_train = np.random.rand(5,3)
y_train = np.array([0,1,0,1,0])
w_tmp = np.random.rand(X_train.shape[1])
b_tmp = 0.5
lambda_tmp = 0.7
dj_dw,dj_db = compute_gradient_linear_reg(X_train,y_train,w_tmp,b_tmp,lambda_tmp)
print(f"Regularized dj_db:{dj_db} and Regularized dj_dw:{dj_dw}")
```

```python
def compute_gradient_logistic_reg(X,y,w,b,lambda_=1):
    m,n=X.shape
    dj_dw = np.zeros(n)
    dj_db=0.
    for i in range(m):
        f_wb_i = sigmoid(np.dot(X[i],w)+b)
        err = f_wb_i - y[i]
        for j in range(n):
            dj_dw[j]+= err*X[i,j]
        dj_db+=err
    dj_dw/=m
    dj_db/=m
    for j in range(n):
        dj_dw[j]+= (lambda_/m)*w[j]
    return dj_dw,dj_db
```

```python
np.random.seed(1)
X_train = np.random.rand(5,3)
y_train = np.array([0,1,0,1,0])
w_tmp = np.random.rand(X_train.shape[1])
b_tmp = 0.5
lambda_tmp = 0.7
dj_dw,dj_db = compute_gradient_logistic_reg(X_train,y_train,w_tmp,b_tmp,lambda_tmp)
print(f"Regularized dj_db:{dj_db} and Regularized dj_dw:{dj_dw}")
```
